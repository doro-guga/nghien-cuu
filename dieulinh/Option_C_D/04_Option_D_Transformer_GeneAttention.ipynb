# Option D: Transformer with Gene-level Attention

This notebook implements a Transformer-based model for binary cancer classification using selected gene expression features.

Each selected gene is treated as a token. The model combines a learnable gene identity embedding with the corresponding expression value projection. Multi-head self-attention is then used to learn global interactions among genes.

```python
import pandas as pd
import numpy as np
from pathlib import Path

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

import matplotlib.pyplot as plt
```

```python
#Cố định seed và chọn GPU
RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
torch.manual_seed(RANDOM_STATE)
torch.cuda.manual_seed_all(RANDOM_STATE)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using device:", device)
```

```python
BASE_DIR = Path("/content/drive/MyDrive/gene_project")

RESULT_DIR = BASE_DIR / "results_gnn"
RESULT_DIR.mkdir(parents=True, exist_ok=True)

X_TRAIN_PATH = BASE_DIR / "X_train_rF_DEG_train_lung.csv"
X_TEST_PATH = BASE_DIR / "X_test_rF_DEG_train_lung.csv"
Y_TRAIN_PATH = BASE_DIR / "y_train_rF_DEG_train.csv"
Y_TEST_PATH = BASE_DIR / "y_test_rF_DEG_train.csv"
```

```python
#Load dữ liệu
X_train_full = pd.read_csv(X_TRAIN_PATH)
X_test = pd.read_csv(X_TEST_PATH)

y_train_full = pd.read_csv(Y_TRAIN_PATH)
y_test = pd.read_csv(Y_TEST_PATH)

print("Raw X_train:", X_train_full.shape)
print("Raw X_test:", X_test.shape)
print("Raw y_train:", y_train_full.shape)
print("Raw y_test:", y_test.shape)
```

Làm sạch dữ liệu

```python
def clean_feature_df(X_df):
    X_df = X_df.copy()
    X_df = X_df.loc[:, ~X_df.columns.str.contains("^Unnamed")]
    return X_df


def clean_label_df(y_df):
    y_df = y_df.copy()
    y_df = y_df.loc[:, ~y_df.columns.str.contains("^Unnamed")]

    # lấy cột cuối làm label
    y_series = y_df.iloc[:, -1]

    return y_series.astype(int)


X_train_full = clean_feature_df(X_train_full)
X_test = clean_feature_df(X_test)

y_train_full = clean_label_df(y_train_full)
y_test = clean_label_df(y_test)

print("Clean X_train:", X_train_full.shape)
print("Clean X_test:", X_test.shape)
print("Clean y_train:", y_train_full.shape)
print("Clean y_test:", y_test.shape)

print("\nTrain label distribution:")
print(y_train_full.value_counts())

print("\nTest label distribution:")
print(y_test.value_counts())
```

```python
#Kiểm tra gene columns
gene_names = X_train_full.columns.tolist()

print("Number of selected genes:", len(gene_names))
print("First 10 genes:")
print(gene_names[:10])
```

```python
#Scale dữ liệu
scaler = StandardScaler()

X_train_full = pd.DataFrame(
    scaler.fit_transform(X_train_full),
    columns=X_train_full.columns
)

X_test = pd.DataFrame(
    scaler.transform(X_test),
    columns=X_test.columns
)

print("After scaling:")
print(X_train_full.head())
```

```python
#Tách validation set
X_train, X_val, y_train, y_val = train_test_split(
    X_train_full,
    y_train_full,
    test_size=0.2,
    stratify=y_train_full,
    random_state=RANDOM_STATE
)

print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)

print("\nTrain labels:")
print(y_train.value_counts())

print("\nValidation labels:")
print(y_val.value_counts())

print("\nTest labels:")
print(y_test.value_counts())
```

```python
#Dataset class
class GeneExpressionDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X.values, dtype=torch.float32)
        self.y = torch.tensor(y.values, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]
```

```python
#DataLoader
BATCH_SIZE = 32

train_loader = DataLoader(
    GeneExpressionDataset(X_train, y_train),
    batch_size=BATCH_SIZE,
    shuffle=True
)

val_loader = DataLoader(
    GeneExpressionDataset(X_val, y_val),
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_loader = DataLoader(
    GeneExpressionDataset(X_test, y_test),
    batch_size=BATCH_SIZE,
    shuffle=False
)
```

**Transformer model**

```python
class GeneTransformerClassifier(nn.Module):
    def __init__(
        self,
        num_genes,
        d_model=64,
        nhead=4,
        num_layers=2,
        dim_feedforward=128,
        dropout=0.3
    ):
        super().__init__()

        self.num_genes = num_genes
        self.d_model = d_model

        # gene identity embedding
        self.gene_embedding = nn.Embedding(num_genes, d_model)

        # expression value projection
        self.expression_projection = nn.Linear(1, d_model)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            activation="gelu",
            batch_first=True
        )

        self.transformer_encoder = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        self.classifier = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, 64),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(64, 1)
        )

    def forward(self, x):
        """
        x shape: [batch_size, num_genes]
        """

        batch_size, num_genes = x.shape

        gene_ids = torch.arange(num_genes, device=x.device)
        gene_ids = gene_ids.unsqueeze(0).expand(batch_size, num_genes)

        gene_emb = self.gene_embedding(gene_ids)
        expr_emb = self.expression_projection(x.unsqueeze(-1))

        h = gene_emb + expr_emb

        h = self.transformer_encoder(h)

        # mean pooling over gene tokens
        sample_embedding = h.mean(dim=1)

        logits = self.classifier(sample_embedding).squeeze(1)

        return logits
```

```python
#Hàm tính metrics
def compute_metrics(y_true, y_prob, threshold=0.5):
    y_pred = (y_prob >= threshold).astype(int)

    acc = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    sn = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)

    try:
        auc = roc_auc_score(y_true, y_prob)
    except:
        auc = 0.0

    cm = confusion_matrix(y_true, y_pred)

    if cm.shape == (2, 2):
        tn, fp, fn, tp = cm.ravel()
        sp = tn / (tn + fp) if (tn + fp) > 0 else 0
    else:
        sp = 0

    ba = (sn + sp) / 2

    return {
        "ACC": acc,
        "Precision": precision,
        "Sensitivity/Sn": sn,
        "Specificity/Sp": sp,
        "F1": f1,
        "AUC": auc,
        "Balanced Accuracy": ba
    }
```

```python
#Hàm train/evaluate 1 epoch
def run_epoch(model, loader, criterion, optimizer=None):
    is_train = optimizer is not None

    if is_train:
        model.train()
    else:
        model.eval()

    total_loss = 0
    all_probs = []
    all_labels = []

    for X_batch, y_batch in loader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        with torch.set_grad_enabled(is_train):
            logits = model(X_batch)
            loss = criterion(logits, y_batch)

            if is_train:
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

        probs = torch.sigmoid(logits).detach().cpu().numpy()

        all_probs.extend(probs)
        all_labels.extend(y_batch.detach().cpu().numpy())

        total_loss += loss.item() * len(X_batch)

    avg_loss = total_loss / len(loader.dataset)

    return avg_loss, np.array(all_labels), np.array(all_probs)
```

```python
#Khởi tạo Transformer
num_genes = X_train.shape[1]

model = GeneTransformerClassifier(
    num_genes=num_genes,
    d_model=64,
    nhead=4,
    num_layers=2,
    dim_feedforward=128,
    dropout=0.3
).to(device)

pos_count = y_train.sum()
neg_count = len(y_train) - pos_count

pos_weight = torch.tensor(
    [neg_count / pos_count],
    dtype=torch.float32
).to(device)

criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=5e-4,
    weight_decay=1e-4
)

print(model)
print("Positive samples:", pos_count)
print("Negative samples:", neg_count)
print("pos_weight:", pos_weight.item())
```

```python
#train
EPOCHS = 100
PATIENCE = 15

best_val_auc = -1
wait = 0
history = []

for epoch in range(1, EPOCHS + 1):
    train_loss, y_train_epoch, p_train_epoch = run_epoch(
        model,
        train_loader,
        criterion,
        optimizer
    )

    val_loss, y_val_epoch, p_val_epoch = run_epoch(
        model,
        val_loader,
        criterion
    )

    train_metrics = compute_metrics(y_train_epoch, p_train_epoch)
    val_metrics = compute_metrics(y_val_epoch, p_val_epoch)

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "val_loss": val_loss,
        "train_auc": train_metrics["AUC"],
        "val_auc": val_metrics["AUC"],
        "val_ba": val_metrics["Balanced Accuracy"]
    })

    print(
        f"Epoch {epoch:03d} | "
        f"train_loss={train_loss:.4f} | "
        f"val_loss={val_loss:.4f} | "
        f"val_auc={val_metrics['AUC']:.4f} | "
        f"val_ba={val_metrics['Balanced Accuracy']:.4f}"
    )

    if val_metrics["AUC"] > best_val_auc:
        best_val_auc = val_metrics["AUC"]
        wait = 0

        torch.save(
            model.state_dict(),
            RESULT_DIR / "best_transformer_model.pt"
        )
    else:
        wait += 1

    if wait >= PATIENCE:
        print("Early stopping triggered.")
        break
```

```python
#Vẽ training curve
history_df = pd.DataFrame(history)

plt.figure(figsize=(8, 5))
plt.plot(history_df["epoch"], history_df["train_loss"], label="Train loss")
plt.plot(history_df["epoch"], history_df["val_loss"], label="Validation loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Transformer Training Curve")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Chọn threshold tốt trên validation
model.load_state_dict(torch.load(RESULT_DIR / "best_transformer_model.pt"))
model.eval()

val_loss, y_val_true, y_val_prob = run_epoch(
    model,
    val_loader,
    criterion
)

fpr, tpr, thresholds = roc_curve(y_val_true, y_val_prob)

best_index = np.argmax(tpr - fpr)
best_threshold = thresholds[best_index]

print("Best threshold:", best_threshold)

val_metrics = compute_metrics(
    y_val_true,
    y_val_prob,
    threshold=best_threshold
)

val_metrics
```

```python
#Test evaluation
test_loss, y_test_true, y_test_prob = run_epoch(
    model,
    test_loader,
    criterion
)

test_metrics = compute_metrics(
    y_test_true,
    y_test_prob,
    threshold=best_threshold
)

print("Transformer Test Metrics")
print("=" * 50)

for metric, value in test_metrics.items():
    print(f"{metric}: {value:.6f}")

y_test_pred = (y_test_prob >= best_threshold).astype(int)

print("\nClassification Report:")
print(classification_report(y_test_true, y_test_pred, digits=6))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_true, y_test_pred))
```

```python
pd.DataFrame(history).to_csv(
    RESULT_DIR / "transformer_training_history.csv",
    index=False
)

pd.DataFrame([test_metrics]).to_csv(
    RESULT_DIR / "transformer_test_metrics.csv",
    index=False
)

pd.DataFrame({
    "y_true": y_test_true,
    "y_prob": y_test_prob,
    "y_pred": y_test_pred
}).to_csv(
    RESULT_DIR / "transformer_test_predictions.csv",
    index=False
)

with open(RESULT_DIR / "transformer_best_threshold.txt", "w") as f:
    f.write(str(best_threshold))

print("Transformer results saved to:", RESULT_DIR)
```
